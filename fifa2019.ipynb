## Univariate Analyiss

```python
plt.figure(figsize=(10,6))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.countplot(x='Preferred Foot', data=df)
plt.title('Preferred Foot Count')
plt.show()
```

```python
df['Preferred Foot'].value_counts()
```

## Bivariate Analysis

```python
# Age vs Overall Rating
plt.figure(figsize=(10,6))
sns.scatterplot(x='Age',y='Overall', data=df)
plt.title('Age vs Overall Rating')
plt.show()
```

```python
numeric_df = df.select_dtypes(include=['int64','float64'])
corr = numeric_df.corr()
plt.figure(figsize=(50,30))
sns.heatmap(corr, cmap='coolwarm', annot=True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
top_players = df[['Name', 'Overall']].sort_values(
    by='Overall',
    ascending=False
).head(10)
```

```python
top_players
```

```python
club_rating = df.groupby('Club')['Overall'].mean()
club_rating = club_rating.sort_values(ascending=False)
club_rating.head(10)
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x=df['Wage'])
plt.title('Wage Outlier')
plt.xlabel('Wage')
plt.show()
```

```python
df['Position'].value_counts().head(15).plot(kind='bar')
plt.title('Top player positions')
plt.xlabel('Position')
plt.ylabel('Count')
plt.show()
```
