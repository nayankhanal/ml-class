```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/kaggle/input/datasets/abdoomoh/all-seaborn-built-in-datasets/Seaborn All Built-in Datasets/healthexp.csv')
```

```python
df.head()
```

```python
df['Year'].min(), df['Year'].max()
```

```python
df.groupby('Country')['Life_Expectancy'].mean()
```

```python
yearly_trend = df.groupby('Year')[['Spending_USD','Life_Expectancy']].mean()

yearly_trend.plot(figsize=(10,5))
plt.title('Global Trend')
plt.show()
```

```python
sns.histplot(x=df['Life_Expectancy'], bins=10, kde=True)
plt.show()
```

```python
sns.scatterplot(x='Spending_USD', y='Life_Expectancy', data=df)
plt.title('Spending vs Life Expectancy')
plt.show()
```

```python
sns.scatterplot(x='Spending_USD', y='Life_Expectancy', hue='Country', data=df)
plt.title('Spending vs Life Expectancy')
plt.show()
```

```python
sns.heatmap(df[['Spending_USD','Life_Expectancy']].corr())
plt.show()
```

```python
df['Spending_USD'].corr(df['Life_Expectancy'])
```

```python
sns.lineplot(x='Year', y='Life_Expectancy', hue='Country', data=df)
plt.show()
```
