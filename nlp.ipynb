```python
import nltk
nltk.download('all')
```

```python
corpus = """Hello welcome to NLP session by Nayan Khanal.
You can watch course recording from revision.
You can become expert in NLP
"""
```

```python
print(corpus)
```

```python
type(corpus)
```

```python
# Sentence tokenization
from nltk.tokenize import sent_tokenize
documents = sent_tokenize(corpus)
documents
```

```python
# Word tokenization
from nltk.tokenize import word_tokenize
words = word_tokenize(corpus)
words
```

```python
for sentence in documents:
  print(word_tokenize(corpus))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize("Hello Welcome, How're you?")
```

```python
# Treebank Word Tokenizer

from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```

# Stemming

```python
# Porter Stemmer
# Regexp Stemmer
# Snowball Stemmer
```

```python
words = ['eating','eats','eaten','writing','writes','programming','programs','history','finally']
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
for word in words:
  print(word + '------>' + stemmer.stem(word))
```

```python
stemmer.stem('congratulations')
```

```python
stemmer.stem('sitting')
```

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
stemmer.stem('eating')
```

```python
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer('english')
```

```python
for word in words:
  print(word + '------->' + snowball_stemmer.stem(word))
```

```python
snowball_stemmer.stem('goes')
```

# Lemmatizer

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

'''
POS -
Noun - n
verb - v,
adjective - a
adverb - r
'''
```

```python
lemmatizer.lemmatize('going', pos='v')
```

```python
lemmatizer.lemmatize('fairly', pos='r')
```

```python
lemmatizer.lemmatize('sportingly', pos='r')
```
