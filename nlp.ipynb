```python
for i in range(len(sentences)):
  words = word_tokenize(sentences[i])
  words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
  sentences[i] = ' '.join(words)
```

```python
for i in range(len(sentences)):
  words = nltk.word_tokenize(sentences[i])
  words = [lemmatizer.lemmatize(word.lower(), pos='v') for word in words if word not in set(stopwords.words('english'))]
  sentences[i] = ' '.join(words)
```

# Text Vectorization

### Bag of Words (BOW)


I love NLP because I love AI

Vocabulary: -----> I love NLP because AI

Word.           Count
I.                2
love.             2
NLP.              1
because           1
AI.               1

BOW vector ---> [2, 2, 1, 1, 1]



### TF-IDF

TF - How much word is important?

TF(love) = 2/7
TF(NLP) = 1/7


IDF - How rare is the word in the whole dataset?

Total 1000 documents
the word appearing in 990

IDF = log(1000/990)

another word appears in only 10

IDF = log(1000/10)

Main Formula = TF * IDF


Dogs love playing

Total words = 3
Word Dogs appears 1

TF = 1/3 = 0.333

Suppose Dogs appears in 50 documents.
Total 100 documents

IDF = log(100/50) = 0.6

TF-IDF = 0.3333 * 0.6
