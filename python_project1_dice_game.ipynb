```python
import random

dice = [1,2,3,4,5,6]

def roll_dice():
  print(random.choice(dice))

while True:
  command = input("Type 'r' to roll the dice and 'q' to quit the game: ")
  if command == 'r':
    res = roll_dice()
  elif command == 'q':
    break
```
