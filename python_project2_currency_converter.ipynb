```python
import requests

def currency_converter(from_currency, amount, to_currency):
  try:
    if type(amount) != 'int':
      amount = float(amount)

    response = requests.get(f'https://api.fastforex.io/fetch-one?from={from_currency}&to={to_currency}&api_key=4e97ca78d1-16f1c045cb-tesa9w')

    formatted_response = response.json()

    if 'error' in formatted_response:
      print(formatted_response['error'])
      return

    to_be_multiplied_amount = formatted_response['result'][to_currency]

    result = amount*to_be_multiplied_amount
    print(result)
  except Exception as e:
    print(e)


from_currency = input("Enter from which currency: ")
amount = input("Amount: ")
to_currency = input("Enter to which currency: ")

currency_converter(from_currency, amount, to_currency)

```
