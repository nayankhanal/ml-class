```python
import random

winning_patterns = [
    [1, 2, 3],  # top row
    [4, 5, 6],  # middle row
    [7, 8, 9],  # bottom row
    [1, 4, 7],  # left column
    [2, 5, 8],  # middle column
    [3, 6, 9],  # right column
    [1, 5, 9],  # diagonal top-left to bottom-right
    [3, 5, 7],  # diagonal top-right to bottom-left
]

computer_chosen = []
user_chosen = []

def choose():
  chosen_fields = computer_chosen + user_chosen
  remaining_fields = [x for x in range(1,10) if x not in chosen_fields]
  # print("chosen: ", chosen_fields)
  # print("remaining: ", remaining_fields)

  if len(remaining_fields) == 0:
    print("Game over!")
    return False

  chosen = input(f"Choose one place from ({remaining_fields}): ")

  try:
    chosen = int(chosen)
  except Exception as e:
    print(e)
    return True

  if chosen in chosen_fields:
    print("This field is already chosen.")
    return True

  if chosen not in remaining_fields:
    print("Invalid field")
    return True

  user_chosen.append(chosen)
  chosen_fields.append(chosen)
  remaining_fields.remove(chosen)
  print("You choose: ",user_chosen)

  if len(user_chosen) > 2:
    for li in winning_patterns:
      if all([pos in user_chosen for pos in li]):
        print("You win!")
        return False

  if len(remaining_fields) == 0:
    print("Game over!")
    return False

  computer = random.choice(remaining_fields)

  computer_chosen.append(computer)
  chosen_fields.append(computer)
  remaining_fields.remove(computer)
  print("Computer choose: ",computer_chosen)

  if len(computer_chosen) > 2:
    for li in winning_patterns:
      if all([pos in computer_chosen for pos in li]):
        print("Computer wins!")
        return False

while True:
  res = choose()

  if res == False:
    break
```
