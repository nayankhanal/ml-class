```python
import random

four_digit = random.choices(range(1,10), k=4)

while True:

  bulls = 0
  cows = 0

  num = input("Guess the 4 digit number: ")
  if type(num) != 'int':
    try:
      num = int(num)
    except Exception as e:
      print(e)
      continue

  num = list(str(num))
  num = [int(i) for i in num]

  if four_digit == num:
    print("You win!")
    break

  for index,value in enumerate(num):
    if value in four_digit and four_digit[index] == value:
      bulls +=1
    elif value in four_digit:
      cows +=1

  print("Bulls: ", bulls)
  print("Cows: ", cows)
```
