```python
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):

    # def __init__(self, num_features):
    #     super().__init__()
    #     self.linear1 = nn.Linear(num_features, 3)
    #     self.relu = nn.ReLU()
    #     self.linear2 = nn.Linear(3, 1)
    #     self.sigmoid = nn.Sigmoid()

    # lets make this sequential
    def __init__(self, num_features):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(num_features, 3),
            nn.ReLU(),
            nn.Linear(3, 1),
            nn.Sigmoid()
        )

    # def forward(self, features):
    #     out = self.linear1(features)
    #     out = self.relu(out)
    #     out = self.linear2(out)
    #     out = self.sigmoid(out)
    #     return out

    # this too
    def forward(self, features):
        out = self.network(features)
        return out
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# model.forward(features) , instead of doing this pytorch encourages you to call directly on model instead of doing .forward() even though you are calling forward method
model(features)
```
