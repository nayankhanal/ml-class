```python
# import torch
import time

# define size of the matrices
size = 1000

# create random matrices
matrix_cpu1 = torch.rand(size, size)
matrix_cpu2 = torch.rand(size, size)

# calculate time of their multiplication
start_time = time.time()
result_cpu = torch.matmul(matrix_cpu1, matrix_cpu2) # matrix multiplication
cpu_time = time.time() - start_time

print("Time taken by the CPU to multiply the matrices: ", cpu_time)

# create random matrices
matrix_gpu1 = matrix_cpu1.to(device)
matrix_gpu2 = matrix_cpu2.to(device)

# calculate time of their multiplication
start_time = time.time()
result_gpu = torch.matmul(matrix_gpu1, matrix_gpu2) # matrix multiplication
torch.cuda.synchronize() # ensure all GPU operations are complete
gpu_time = time.time() - start_time

print("Time taken by the GPU to multiply the matrices: ", gpu_time)
```
