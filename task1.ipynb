# Arithmetic with Mixed Types


*   Create one int variable (a) and one float variable (b).
*   Print the sum, difference, product, and quotient of a and b.
*   Print the type() of each result (notice how types may change).

```python
a = 7
b = 5.5

sum = a+b
diff = a-b
prod = a*b
div = a%b

print(sum, type(sum))
print(diff, type(diff))
print(prod, type(prod))
print(div, type(div))
```

# In-Place vs. Reassignment with String Methods

*   Let word = "example".
*   Call word.upper() but do not store it, then print word.
*   Now set word = word.upper(), then print word.
*   Comment on why the second print is different from the first.





```python
word = "example"
print(word.upper())
word = word.upper()
print(word)

# second print is different from the first because first is not assigned to word and second one is assigned to the variable word,
# actually stored in different memory location as word is immutable
```
