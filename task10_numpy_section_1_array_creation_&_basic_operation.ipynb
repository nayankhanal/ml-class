# Section 1: Array Creation & Basic Properties (1-30)

Import the NumPy library and print its version.

```python
import numpy as np

print(np.__version__)
```

Create a NumPy array containing numbers from 1 to 10.

```python
np_array = np.array(range(1,11))
print(np_array)
```

Create an array of all zeros with size 10.

```python
zeros = np.zeros(10)
print(zeros)
```

Create an array of all ones with shape (3,3).

```python
ones = np.ones(10)
print(ones)
```

Create an identity matrix of size 5×5.

```python
iden1 = np.eye(N=5, M=None, k=0, dtype=float)
iden2 = np.identity(5)

print(iden1)
print(iden2)
```

Create an array of even numbers from 2 to 50.

```python
even_array = np.arange(2,52,2)
print(even_array)
```

Create an array of odd numbers from 1 to 99.

```python
odd_array = np.arange(1,100,2)
print(odd_array)
```

Generate numbers from 10 to 100 with step size 5.

```python
nums = np.arange(10, 101, 5)
print(nums)
```

Create a NumPy array from a Python list.

```python
arr = [1,2,3,4,5]
np_arr = np.array(arr)
print(np_arr)
```

Create a NumPy array from a tuple

```python
arr = (1,2,3,4,5)
tpl_arr = np.array(arr)
print(np_arr)
```

Find the shape of an array.

```python
array = np.array([2,35,25,2,4])
array.shape
```

Find the dimensions of an array.

```python
array.ndim
```

Find the datatype of array elements

```python
print(type(array))
print(type(array[0]))
```

Generate 10 random integers between 1 and 100.

```python


random_arr = np.random.randint(1,101,size=10)
print(random_arr)
```

Reshape an array of size 12 into (3,4).

```python
reshaped_arr = np.arange(1,13).reshape(3,4)
print(reshaped_arr)
```

Flatten a 3×3 matrix into a 1D array.

```python
three_three = np.arange(1,10).reshape(3,3)
print(three_three)

flat = three_three.flatten()
print(flat)
```

Create a 4×4 matrix filled with value 7.

```python
four_four = np.full((4,4),7)
print(four_four)
```

Find the size (number of elements) of an array.

```python
four_four.size
```

Create a diagonal matrix with diagonal values [1,2,3,4].

```python
diagonal = [1,2,3,4]

diag_matrix = np.diag(diagonal)

print(diag_matrix)
```

Convert a float array into an integer array.

```python
float_arr = np.array([1.9, 2.1, 3.7, 4.5])
int_arr = float_arr.astype(int)
print(int_arr)
```

Create an array with values between 0 and 1 using linspace().

```python
lin_arr = np.linspace(0,1,10)
print(lin_arr)
```

Reverse an array.

```python
print(lin_arr[::-1])
print(np.flip(lin_arr))
```

Sort an array in ascending order.

```python
arr = [21,2,5,7,1,65,56]

print(np.sort(arr))
```

Find the maximum and minimum values in an array.

```python
print(np.min(arr))
print(np.max(arr))
```

Find the index of maximum value.

```python
print(np.argmax(arr))
```

Find the index of minimum value.

```python
print(np.argmin(arr))
```

Create a 2×3 matrix and print its transpose.

```python
mtrx = np.matrix(range(1,7)).reshape(2,3)
print(mtrx)
print(mtrx.transpose())
```

Concatenate two arrays.

```python
mtrx2 = np.array(range(1,7))
mtrx3 = np.array(range(7,13))

concated = np.concatenate((mtrx2, mtrx3))

print(concated)
```

Split an array into 3 equal parts.

```python
parts = np.split(concated,4)
print(parts)
```
