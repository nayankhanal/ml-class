# Section 2: Basic Operations (31-50)

Add two arrays element-wise.

```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4])
arr2 = np.array([3,4,5,6])

print(arr1 + arr2)
```

Subtract two arrays element-wise.

```python
print(arr2 - arr1)
```

Multiply two arrays element-wise

```python
print(arr1*arr2)
```

Divide two arrays element-wise.

```python
print(arr1/arr2)
```

Find square of every element in an array

```python
print(np.square(arr1))
```

Find cube of every element.

```python
print(np.pow(arr1,3))
```

Calculate square root of array values.

```python
print(np.sqrt(np.square(arr1)))
```

Find exponential values of array elements.

```python
print(np.exp(arr1))
```

Find logarithm of array elements

```python
print(np.log(arr1))
```

Compute sum of all elements in an array.

```python
print(np.sum(arr1))
```

Compute mean of array values.

```python
print(np.mean(arr1))
```

Compute median of array values.

```python
print(np.median(arr1))
```

Compute standard deviation.

```python
print(np.std(arr1))
```

Compute variance.

```python
print(np.var(arr1))
```

Find cumulative sum of elements.

```python
print(np.cumsum(arr1))
```

Find cumulative product of elements.

```python
print(np.cumprod(arr1))
```

Find product of all elements in array.

```python
print(np.prod(arr1))
```

Round decimal values to 2 decimal places.

```python
arr = np.array([1.23456, 4.8972, 9.001, 16.555, 25.1])

print(np.round(arr,decimals=2))
```

Find absolute values of negative numbers.

```python
arr3 = [1,2,-4,6,-7]

print(np.abs(arr3))
```

Replace all negative values with zero.

```python
print(np.maximum(0, arr3))
```
