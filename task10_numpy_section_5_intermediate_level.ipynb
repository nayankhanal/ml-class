# Section 5: Intermediate Level (81-90)

Generate a random array and normalize it between 0 and 1.

```python
import numpy as np
```

```python
array = np.array(range(0,9))

min = np.min(array)
max = np.max(array)

normalized_array = (array - min) / (max - min)

print(normalized_array)
```

Create a checkerboard pattern matrix (8×8).

```python
checkerboard = np.zeros((8,8), dtype=int)

# print(checkerboard)

checkerboard[1::2,::2] = 1
checkerboard[::2, 1::2] = 1

print(checkerboard)
```

Count frequency of each unique value.

```python
array2 = np.random.randint(0,21,10)
print(array2)

values, counts = np.unique(array2, return_counts=True)

frequencies = dict(zip(values.tolist(),counts.tolist()))

print(frequencies)
```

Shuffle elements of an array randomly.

```python
array3 = np.random.randint(0,21,10)
print(array3)

rng = np.random.default_rng()

# creates copy
shuffled_copy = rng.permutation(array3)

# modifies same array
rng.shuffle(array3)

print(array3)
print(shuffled_copy)
```

Find common elements between two arrays.

```python
array4 = np.random.randint(0,50,10)
array5 = np.random.randint(0,50,10)

print(array4)
print(array5)

print(np.intersect1d(array4, array5))
```

Remove duplicate values from an array.

```python
# sorted one

array6 = np.random.randint(0,50,10)
print(array6)

sorted_unique_arr = np.unique(array6)
print(sorted_unique_arr)
```

```python
# without sorting

array6 = np.random.randint(0,50,10)
print(array6)

values, indices = np.unique(array6, return_index=True)

ordered_unique_arr = values[np.argsort(indices)]

print(ordered_unique_arr)
```

Find missing numbers in an array from 1-100.

```python
ten_to_100_all = np.arange(1,101)

# ten_to_100_with_missing = np.random.randint(1,101,97)
ten_to_100_with_missing = np.random.choice(ten_to_100_all, size=97, replace=False)

missing_values = np.setdiff1d(ten_to_100_all, ten_to_100_with_missing)

print(missing_values)
```

Convert a 1D array into a 3D array.

```python
array1d = np.arange(0,16)
print(array1d)

print(array1d.reshape(2,4,2))
```

Create a border matrix where borders are 1 and inner values are 0.

```python
border_matrix = np.ones((8,8), dtype=int)

border_matrix[1:-1:1, 1:-1:1] = 0
print(border_matrix)
```
