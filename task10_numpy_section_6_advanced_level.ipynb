# Section 6: Advanced Level (91-100)

Implement one-hot encoding using NumPy only.

```python
import numpy as np
```

```python
labels = np.array([0, 2, 1, 0, 2])
num_classes = 3

print(np.eye(num_classes, dtype=int))
print("\n")
onehot = np.eye(num_classes, dtype=int)[labels]

print(onehot)
```
