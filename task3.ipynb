### String Formatting
* Create two variables: `name = "Rajesh"` and `score = 95`.
* Print: "Alice scored 95 points." using two methods (concatenation and an f-string or `str.format()`).

```python
name = "Rajesh"
score = 95

def fstring(name, score):
  return print(f"{name} scored {score} points.")

def concat(name, score):
  return print(name + " scored " + str(score) + " points.")


fstring(name, score)
concat(name, score)

printable = "{name} scored {score} points."

print(printable.format(name = name, score = score))
```

### Casefold vs. Lower
* **Create two strings**, `"Case"` and `"case"`.
* **Compare them** with the regular `==` operator directly.
* **Compare them again** after applying `.casefold()`.
* **Print results** and comment on how `.casefold()` differs from `.lower()` in edge cases.

```python
cs1 = "Case"
cs2 = "case"

print(cs1 == cs2)

print(cs1.casefold(), cs2.casefold())

print(cs1.casefold() == cs2.casefold())

# .lower() standard for showing to users, .casefold() for actual comparision,
# .lower() converts A-Z to a-z; .casefold() is "stronger" and converts
# international characters (like German 'ß' to 'ss') for accurate matching.
```
