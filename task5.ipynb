### Iterate Through a Tuple and Print Types
*   **Task:** Create a tuple with at least 5 different types of elements (`int`, `float`, `string`, `bool`, and `complex`).
    *   **Use a `for` loop** to iterate over the tuple.
    *   **Print each element** along with its data type using the `type()` function.

```python
tpl = (1,2.5,"nayan",True,2+4j)

for i in tpl:
  print(i, type(i))
```

### Print All Items from a List with a Custom Message
*   **Task:** Create a list of 4 different city names.

### Print Each Character of a String with Its Index
*   **Task:** Ask the user to enter a word.
    *   **Use a `for` loop and `enumerate()`** to iterate through the string.
    *   **Print each character** along with its corresponding index position.

```python
word = input("Enter a word")

for i,w in enumerate(word):
  print(i, w)
```

### Sum of Elements in a List
*   **Task:** Create a list of integers.
    *   **Use a `for` loop** to calculate the sum of all the elements.
    *   **Print the total** sum after the loop finishes.

```python
lst2 = [1,2,3,4,5,6,7]
sum = 0

for i in lst2:
  sum += i

print(sum)
```

### Check for Vowels in a String
*   **Task:** Ask the user to enter a word.
    *   **Use a `for` loop** to iterate through each character of the word.
    *   **Check if the character is a vowel** (a, e, i, o, u).
    *   **Print a message** for each character that is a vowel (e.g., "Found a vowel: e").

```python
wrd = input("Enter a word ")

for w in wrd:
  if w in ('a','e','i','o','u'):
    print("Found a vowel: ",w)
```

### Count Numbers Greater Than 10 in a List
*   **Task:** Create a list of integers.
    *   **Use a `for` loop** to iterate through the list.
    *   **Check if each number** is greater than 10.
    *   **Print the final count** of numbers that met the condition.
  

```python
lst = [1,6,3,56,23,12,3,7,31,13]
count = 0

for l in lst:
  if l > 10:
    count += 1

print(count)
```
