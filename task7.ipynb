### Function with Default Argument
*   **Task:** Create a function `greet(name, message="Welcome!")` that prints:
    *   `"Hello, <name>! <message>"`
*   **Requirements:**
    *   Call the function **with** both arguments.
    *   Call the function **without** the `message` argument to see the default value in action.
  

```python
def greet(name, message="Welcome!"):
  print(f"Hello, {name}! {message}")

greet("Nayan", "K xa")
greet("Elon")
```

### Function with *args (Variable Number of Arguments)
*   **Task:** Create a function `total(*numbers)` that takes any number of numeric arguments and returns their sum.
*   **Requirements:**
    *   The function should handle an arbitrary number of inputs.
    *   **Example Call:** `total(5, 10, 15)` should return `30`.

```python
def total(*numbers):
  return sum(numbers)

total(5, 10, 15)
```

### Function with **kwargs (Keyword Arguments)
*   **Task:** Create a function `display_info(**details)` that accepts an arbitrary number of keyword arguments.
    *   **Logic:** Use a loop to iterate through the dictionary of arguments.
    *   **Output:** Print each key and value pair in a formatted way.
*   **Expected Example Output:**
    *   `name: Rajesh`
    *   `age: 25`
    *   `city: Kathmandu`

```python
def display_info(**details):
  for key,value in details.items():
    print(f"{key}: {value}")

display_info(name= "Rajesh", age= 25, city= "Kathmandu")
```

### Function Using Both *args and a Default Argument
*   **Task:** Create a function `repeat_words(*words, times=2)` that repeats each provided word a specific number of times.
    *   **Arguments:** Use `*words` to collect multiple strings and `times` as a keyword argument with a default value of 2.
    *   **Logic:** Iterate through the `words` and print each one multiplied by the `times` value.
*   **Example Call:** `repeat_words("Hi", "Bye", times=3)`
*   **Expected Output:**
    *   `Hi Hi Hi`
    *   `Bye Bye Bye`

```python
def repeat_words(*words, times=2):
  for w in words:
    print((w + " ") * times)

repeat_words("Hi", "Bye", times=3)
```

### Function Using All Types of Parameters Together
*   **Task:** Create a function `user_profile(name, age=18, *hobbies, **extra_info)` that demonstrates the use of positional, default, variable, and keyword arguments.
*   **Logic:**
    *   **Print Name and Age:** Display the required `name` and the defaultable `age`.
    *   **Handle Hobbies:** Use a loop to iterate through and print all items captured by `*hobbies`.
    *   **Handle Extra Info:** Use a loop to print any additional key-value pairs captured by `**extra_info`.
*   **Requirements:**
    *   Test the function by passing various numbers of hobbies and extra details (like `location` or `job`) to see how the arguments are distributed.

```python
def user_profile(name, age=18, *hobbies, **extra_info):
  print("Name: ", name)
  print("Age: ", age)

  for h in hobbies:
    print(h)

  for key,value in extra_info.items():
    print(f"{key}: {value}")


user_profile("Ram", 21, "Football", "Reading", "Painting", address="KTM", education="+2")
```
