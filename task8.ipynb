### 2. Polymorphism
*   **Task:** Demonstrate how different classes can be treated as instances of the same general class through a single interface.
    *   **Shared Method Name:** Define multiple classes (e.g., `Dog` and `Cat`) that both contain a method with the exact same name, such as `details()`.
    *   **Polymorphic Function:** Create a standalone function `show_details(obj)` that accepts an object as an argument and calls `obj.details()`.
    *   **Execution:** Instantiate objects from your different classes and pass them into `show_details()` to see how the same function call produces different results depending on the object type.
*   **Goal:** Understand how one interface (the function) can be used for a general class of actions across different data types.

```python
class Dog:
  def details(self):
    print("Dog class")

class Cat:
  def details(self):
    print("Cat class")

def show_details(obj):
  obj.details()


dog = Dog()
cat = Cat()

show_details(dog)
show_details(cat)
```

### 3. Single Inheritance
*   **Task:** Create a parent and child class relationship to demonstrate basic inheritance.
    *   **Base Class (`Vehicle`):** Define a class with a method called `start()` that prints a general message (e.g., "Vehicle is starting").
    *   **Derived Class (`Bike`):** Create this class and ensure it inherits from `Vehicle`.
    *   **Unique Method:** Add a method to the `Bike` class called `wheels()` that prints "Two wheels".
*   **Requirements:**
    *   Instantiate an object of the `Bike` class.
    *   Call the inherited `start()` method and the class-specific `wheels()` method to show that the child class has access to both.

---



*   **Goal:** Understand how a child class (subclass) automatically acquires the attributes and methods of the parent class (superclass), promoting code reusability.

```python
class Vehicle:
  def start(self):
    print("Vehicle is starting")

class Bike(Vehicle):
  def wheels(self):
    print("Two wheels")

bike = Bike()
bike.start()
bike.wheels()
```

### 6. Polymorphism Using a Loop
*   **Task:** Create classes `Pen` and `Pencil`, both containing a method named `write()`.
*   **Requirements:**
    *   Store instances of both `Pen` and `Pencil` in a single list.
    *   Iterate through the list using a loop.
    *   Call the `write()` method on each object to demonstrate polymorphic behavior.

### 7. Use Encapsulation to Protect Age
*   **Task:** Create a class `Person` with a private attribute `__age`.
*   **Requirements:**
    *   Implement a method `set_age()` to set the age value.
    *   **Validation:** If the age is less than 0, print "Invalid age".
    *   Implement a method `get_age()` to return the current age.

```python
class Person:
  __age = 16

  def set_age(self, age):
    if age < 0:
      print("Invalid age")
      return
    self.__age = age

  def get_age(self):
    return self.__age

person = Person()
person.set_age(43)
person.get_age()
```

### 8. Company Structure

### 12. Get Length of Each Word

### 14. Create a Word by Joining Letters
*   **Task:** Given a list of letters `['P', 'Y', 'T', 'H', 'O', 'N']`, combine them into a single string.
*   **Requirements:**
    *   Use the `reduce()` function from the `functools` module.
    *   Use a `lambda` function to concatenate the letters into the word "PYTHON".

```python
from functools import reduce

letters = ['P', 'Y', 'T', 'H', 'O', 'N']

concated = reduce(lambda x,y: x+y, letters)
print(concated)
```
