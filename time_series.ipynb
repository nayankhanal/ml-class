```python
df['date'] = pd.to_datetime(df['date'])
df['date']
```

```python
df['date'].dt.year
```

```python
df = df.sort_values('date')
df
```

```python
df.sample()
```

```python
df['month'] = df['date'].dt.month
df.sample()
```

```python
df['day'] = df['date'].dt.day
df.sample()
```

```python
df['year'] = df['date'].dt.year
df.sample()
```

```python
df['dayofweek'] = df['date'].dt.dayofweek # Mon = 0, Tues = 1 ...... Sun = 6
df.sample()
```

```python
df['weekofyear'] = df['date'].dt.isocalendar().week
df.sample(5)
```

```python
df['quarter'] = df['date'].dt.quarter
df.sample(5)
```

```python
df['is_weekend'] = (df['dayofweek'] >=5 ).astype(int)
df.sample(5)
```

```python
# Lag Features
```

```python
df.head()
```

```python
df['lag_1'] = df['sales'].shift(1)
df.head(5)
```

```python
df['lag_7'] = df['sales'].shift(7)
df.head(8)
```

```python
df['sales'].shift(1).rolling(3).mean()
```

```python
df['diff_1'] = df['sales'].diff(1)
df.head()
```

```python
df['pct_change'] = df['sales'].pct_change() * 100
df.head()
```
